# Mini-Proyecto: Explorador de Directorios

Este mini proyecto simula un explorador de archivos. El sistema de archivos lo representamos con un diccionario anidado: si el valor es otro diccionario, es una **carpeta**; si es un texto, es un **archivo** con su tamaño.

```python
sistema_archivos = {
    "Documentos": {
        "apuntes.txt": "10kb",
        "fotos": {
            "vacaciones.png": "2mb",
            "perro.jpg": "1mb"
        }
    },
    "sistema.dll": "500kb"
}
```

## Clase `ExploradorRecursivo`

- **`imprimir_arbol`**: muestra la estructura con sangrías. Si es una carpeta, imprime su nombre y se vuelve a llamar con `nivel + 1`. Si es un archivo, lo imprime con su tamaño.
- **`buscar_archivo`**: busca la ruta de un archivo. Si es una carpeta, se llama a sí mismo adentro de ella y, si lo encuentra, le pega el nombre de la carpeta adelante. Si no lo encuentra, devuelve `None`.
- **`contar_archivos`**: cuenta los archivos (las carpetas no cuentan). Si es una carpeta, suma los archivos que tiene adentro. Si es un archivo, suma 1.

```python
class ExploradorRecursivo:

    def imprimir_arbol(self, sistema, nivel=0):
        for nombre, contenido in sistema.items():
            if isinstance(contenido, dict):
                print("    " * nivel + nombre + "/")
                self.imprimir_arbol(contenido, nivel + 1)
            else:
                print("    " * nivel + nombre + " (" + contenido + ")")

    def buscar_archivo(self, sistema, nombre):
        for clave, contenido in sistema.items():
            if isinstance(contenido, dict):
                ruta = self.buscar_archivo(contenido, nombre)
                if ruta is not None:
                    return clave + "/" + ruta
            elif clave == nombre:
                return clave
        return None

    def contar_archivos(self, sistema):
        total = 0
        for contenido in sistema.values():
            if isinstance(contenido, dict):
                total += self.contar_archivos(contenido)
            else:
                total += 1
        return total
```

## Pruebas

```python
explorador = ExploradorRecursivo()
explorador.imprimir_arbol(sistema_archivos)
```

```python
print(explorador.buscar_archivo(sistema_archivos, "perro.jpg"))
print(explorador.buscar_archivo(sistema_archivos, "sistema.dll"))
print(explorador.buscar_archivo(sistema_archivos, "nada.txt"))
```

```python
print(explorador.contar_archivos(sistema_archivos))
```
